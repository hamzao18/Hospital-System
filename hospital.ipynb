```python
class Patient:
    def __init__(self, patient_id, name, age, gender, contact):
        self.patient_id = patient_id
        self.name = name
        self.age = age
        self.gender = gender
        self.contact = contact

class Doctor:
    def __init__(self, doctor_id, name, specialization, contact):
        self.doctor_id = doctor_id
        self.name = name
        self.specialization = specialization
        self.contact = contact

class Appointment:
    def __init__(self, appointment_id, patient_id, doctor_id, date):
        self.appointment_id = appointment_id
        self.patient_id = patient_id
        self.doctor_id = doctor_id
        self.date = date

class Hospital:
    def __init__(self):
        self.patients = []
        self.doctors = []
        self.appointments = []
    def add_patient(self):
     print("\n--- Add Patient ---")
     patient_id = len(self.patients) + 1
     name = input("Enter patient name: ")
     while True:
      age_input = input("Enter patient age : ")
      try:
        age = int(age_input)
        if 0 < age < 120:
         break
        else:
            raise ValueError("Please enter a realistic age between 0 and 120.")
      except ValueError:
        print("Invalid input. Please enter a positive integer.")
     gender = input("Enter patient gender: ")
     contact = input("Enter patient contact: ")
     self.patients.append(Patient(patient_id, name, age, gender, contact))
     print(f"Patient {name.capitalize()} added successfully with ID {patient_id}.")

    def view_patients(self):
        print("\n--- Patient List ---")
        if not self.patients:
            print("No patients found.")
            return
        for patient in self.patients:
            print(f"ID: {patient.patient_id}, Name: {patient.name.title()}, Age: {patient.age}, Gender: {patient.gender.capitalize()}, Contact: {patient.contact}")

    def search_patient(self):
        patient_id = int(input("Enter patient ID to search: "))
        patient = next((p for p in self.patients if p.patient_id == patient_id), None)
        if patient:
            print(f"ID: {patient.patient_id}, Name: {patient.name.title()}, Age: {patient.age}, Gender: {patient.gender.capitalize()}, Contact: {patient.contact}")
        else:
            print("Patient not found.")

    def add_doctor(self):
        print("\n--- Add Doctor ---")
        doctor_id = len(self.doctors) + 1
        name = input("Enter doctor name: ")
        specialization = input("Enter doctor specialization: ")
        contact = input("Enter doctor contact: ")
        self.doctors.append(Doctor(doctor_id, name, specialization, contact))
        print(f"Doctor {name} added successfully with ID {doctor_id}.")

    def view_doctors(self):
        print("\n--- Doctor List ---")
        if not self.doctors:
            print("No doctors found.")
            return
        for doctor in self.doctors:
            print(f"ID: {doctor.doctor_id}, Name: {doctor.name.title()}, Specialization: {doctor.specialization.capitalize()}, Contact: {doctor.contact}")

    def search_doctor(self):
        doctor_id = int(input("Enter doctor ID to search: "))
        doctor = next((d for d in self.doctors if d.doctor_id == doctor_id), None)
        if doctor:
            print(f"Found Doctor - ID: {doctor.doctor_id}, Name: {doctor.name.title()}, Specialization: {doctor.specialization.capitalize()}, Contact: {doctor.contact}")
        else:
            print("Doctor not found.")

    def book_appointment(self):
        print("\n--- Book Appointment ---")
        appointment_id = len(self.appointments) + 1
        patient_id = int(input("Enter patient ID: "))
        doctor_id = int(input("Enter doctor ID: "))
        date = input("Enter appointment date (YYYY-MM-DD): ")

        patient = next((p for p in self.patients if p.patient_id == patient_id), None)
        doctor = next((d for d in self.doctors if d.doctor_id == doctor_id), None)

        if not patient:
            print("Invalid patient ID.")
            return
        if not doctor:
            print("Invalid doctor ID.")
            return

        self.appointments.append(Appointment(appointment_id, patient_id, doctor_id, date))
        print(f"Appointment booked successfully for Patient {patient.name} with Doctor {doctor.name} on {date}.")

    def view_appointments(self):
        print("\n--- Appointment List ---")
        if not self.appointments:
            print("No appointments found.")
            return
        for appointment in self.appointments:
            patient = next(p for p in self.patients if p.patient_id == appointment.patient_id)
            doctor = next(d for d in self.doctors if d.doctor_id == appointment.doctor_id)
            print(f"Appointment ID: {appointment.appointment_id}, Patient: {patient.name.title()}, Doctor: {doctor.name.title()}, Date: {appointment.date}")


    def menu(self):
        while True:
            print("\n--- Hospital Management System ---")
            print("1. Add Patient")
            print("2. View Patients")
            print("3. Search Patient")
            print("4. Add Doctor")
            print("5. View Doctors")
            print("6. Search Doctor")
            print("7. Book Appointment")
            print("8. View Appointments")
            print("9. Exit")
            choice = input("Enter your choice: ")

            if choice == "1":
                self.add_patient()
            elif choice == "2":
                self.view_patients()
            elif choice == "3":
                self.search_patient()
            elif choice == "4":
                self.add_doctor()
            elif choice == "5":
                self.view_doctors()
            elif choice == "6":
                self.search_doctor()
            elif choice == "7":
                self.book_appointment()
            elif choice == "8":
                self.view_appointments()
            elif choice == "9":
                print("Exiting the system. Goodbye!")
                break
            else:
                print("Invalid choice. Please try again.")

if __name__ == "__main__":
    hospital = Hospital()
    hospital.menu()
```
